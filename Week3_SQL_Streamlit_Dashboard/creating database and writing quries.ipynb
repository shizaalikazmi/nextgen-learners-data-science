## Step 3: Create SQLite Database

### Objective
The objective of this step is to create a relational SQLite database using the three prepared datasets: applicants, interns, and hackathon scores.

### Description

In this step, we perform the following operations:

1. Import the `sqlite3` library to work with SQLite databases.
2. Create a new database named `nextgen.db`.
3. Create three relational tables:
   - `applicants`
   - `interns`
   - `hackathon_scores`
4. Define primary keys to uniquely identify records.
5. Define foreign keys to establish relationships between the tables.
6. Insert the data from the three Pandas DataFrames into their respective database tables.

### Database Relationships

- `applicants.applicant_id` is the primary key in the applicants table.
- `interns.intern_id` is the primary key in the interns table.
- `interns.applicant_id` references `applicants.applicant_id`.
- `hackathon_scores.intern_id` references `interns.intern_id`.

### Outcome
A relational SQLite database is created containing all three tables. The database will be used to perform SQL queries, analyze internship performance, and develop the interactive dashboard.

```python
import sqlite3
```

```python
# Create SQLite database
conn = sqlite3.connect("nextgen.db")
```

```python
# Enable foreign key constraints
conn.execute("PRAGMA foreign_keys = ON")
```

```python
# Create Applicants table
conn.execute("""
CREATE TABLE IF NOT EXISTS applicants (
    applicant_id INTEGER PRIMARY KEY,
    "Applicant Name" TEXT,
    Email TEXT,
    Phone TEXT,
    "Domain Applied" TEXT,
    University TEXT,
    "Application Date" TEXT,
    Status TEXT
)
""")
```

```python
# Create Interns table
conn.execute("""
CREATE TABLE IF NOT EXISTS interns (
    intern_id INTEGER PRIMARY KEY,
    applicant_id INTEGER,
    domain TEXT,
    start_date TEXT,
    completion_status TEXT,
    FOREIGN KEY (applicant_id)
    REFERENCES applicants(applicant_id)
)
""")
```

```python
# Create Hackathon Scores table
conn.execute("""
CREATE TABLE IF NOT EXISTS hackathon_scores (
    intern_id INTEGER PRIMARY KEY,
    score INTEGER CHECK(score BETWEEN 0 AND 100),
    domain TEXT,
    FOREIGN KEY (intern_id)
    REFERENCES interns(intern_id)
)
""")
```

```python
# Insert data into database
applicants.to_sql("applicants", conn, if_exists="append", index=False)
interns.to_sql("interns", conn, if_exists="append", index=False)
scores.to_sql("hackathon_scores", conn, if_exists="append", index=False)

conn.commit()

print("SQLite database created successfully!")
```

```python
# Query 1: Counts completed interns grouped by domain

query1 = """
SELECT
    domain,
    COUNT(*) AS completed_count
FROM interns
WHERE completion_status = 'Completed'
GROUP BY domain
ORDER BY completed_count DESC;
"""

result1 = pd.read_sql(query1, conn)

display(result1)
```

### Query 1: Completed Interns per Domain

**Objective:** To calculate how many interns successfully completed the internship program in each domain.

**SQL Concepts Used:**
- `SELECT` – Select the required columns.
- `WHERE` – Filter only completed interns.
- `COUNT(*)` – Count the number of completed interns.
- `GROUP BY` – Group interns according to their domains.
- `ORDER BY` – Sort domains from highest to lowest completion count.

**Expected Outcome:** A result showing each domain and the total number of interns who completed the program in that domain.

```python
# Query 2: Calculates the average hackathon score for each domain

query2 = """
SELECT
    domain,
    ROUND(AVG(score), 2) AS avg_score
FROM hackathon_scores
GROUP BY domain
ORDER BY avg_score DESC;
"""

result2 = pd.read_sql(query2, conn)

display(result2)
```

### Query 2: Average Hackathon Score per Domain

**Objective:** To calculate and compare the average hackathon scores of interns across different domains.

**SQL Concepts Used:**
- `SELECT` – Select the domain and score columns.
- `AVG()` – Calculate the average hackathon score for each domain.
- `ROUND()` – Round the average score to two decimal places.
- `GROUP BY` – Group scores according to their domains.
- `ORDER BY` – Sort domains from highest to lowest average score.

**Expected Outcome:** A result showing each domain and its average hackathon score, helping us identify which domain achieved the highest average performance.

```python
# Query 3: Lists interns who scored 85 or above

query3 = """
SELECT
    i.intern_id,
    i.domain,
    h.score
FROM interns i
JOIN hackathon_scores h
    ON i.intern_id = h.intern_id
WHERE h.score >= 85
ORDER BY h.score DESC;
"""

result3 = pd.read_sql(query3, conn)

display(result3)
```

### Query 3: Identify Top-Performing Interns

**Objective:** To identify interns who achieved a hackathon score of 85 or above and recognize high-performing participants.

**SQL Concepts Used:**
- `SELECT` – Retrieve intern IDs, domains, and hackathon scores.
- `JOIN` – Combine the interns and hackathon_scores tables.
- `ON` – Match records using the common intern_id column.
- `WHERE` – Filter interns who scored 85 or above.
- `ORDER BY` – Arrange interns from highest to lowest score.

**Expected Outcome:** A list of top-performing interns with their intern IDs, domains, and scores, sorted in descending order.

```python
# Query 4: Calculates the applicant-to-completed-intern conversion rate per domain

query4 = """
SELECT
    a."Domain Applied" AS domain,
    COUNT(DISTINCT a.applicant_id) AS total_applicants,
    COUNT(DISTINCT i.intern_id) AS total_completed,
    ROUND(100.0 * COUNT(DISTINCT i.intern_id)
        / COUNT(DISTINCT a.applicant_id), 2) 
    AS conversion_rate_pct
FROM applicants a
LEFT JOIN interns i
    ON a.applicant_id = i.applicant_id
    AND i.completion_status = 'Completed'
GROUP BY a."Domain Applied"
ORDER BY conversion_rate_pct DESC;
"""

result4 = pd.read_sql(query4, conn)

display(result4)
```

### Query 4: Applicant-to-Completed-Intern Conversion Rate per Domain

**Objective:** To calculate the percentage of applicants who successfully completed the internship program in each domain.

**SQL Concepts Used:**
- `SELECT` – Retrieve domain names and applicant/intern counts.
- `LEFT JOIN` – Combine applicants with their corresponding interns while keeping all applicants.
- `ON` – Match applicants and interns using their applicant IDs.
- `COUNT(DISTINCT)` – Count unique applicants and completed interns.
- `ROUND()` – Round the conversion rate to two decimal places.
- `GROUP BY` – Group records according to domains.
- `ORDER BY` – Sort domains according to their conversion rates.

**Expected Outcome:** A result showing each domain's total applicants, completed interns, and conversion rate as a percentage.

```python
# Query 5: Counts completed and dropped-out interns in each domain

query5 = """
SELECT
    domain,
    completion_status,
    COUNT(*) AS total_interns
FROM interns
GROUP BY domain, completion_status
ORDER BY domain, total_interns DESC;
"""

result5 = pd.read_sql(query5, conn)

display(result5)
```
